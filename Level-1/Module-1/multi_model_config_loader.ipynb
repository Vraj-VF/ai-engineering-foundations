```python
from typing import Any

# 1. Configuration Database
MODELS_CONFIG: dict[str, dict[str, Any]] = {
    "gpt-4o": {
        "provider": "OpenAI",
        "max_tokens": 4096,
        "temperature": 0.7,
    },
    "gpt-4o-mini": {
        "provider": "OpenAI",
        "max_tokens": 16384,
        "temperature": 0.5,
    },
    "claude-3-5-sonnet": {
        "provider": "Anthropic",
        "max_tokens": 8192,
        "temperature": 0.3,
    },
    "gemini-1.5-pro": {
        "provider": "Google",
        "max_tokens": 32768,
        "temperature": 1.0,
    },
    "mistral-large": {
        "provider": "Mistral",
        "max_tokens": 2048,
        "temperature": 0.7,
    },
}

def search_by_provider(
    configs: dict[str, dict[str, Any]], provider: str) -> dict[str, dict[str, Any]]:
    """
    Finds all models belonging to a given provider.
    it will handles case-insensitivity and extra whitespace and 
    Returns an empty dict if no matching provider is found.
    """
    clean_target = provider.strip().lower()
    return {
        model_name: config
        for model_name, config in configs.items()
        if config["provider"].strip().lower() == clean_target
    }


def filter_by_max_tokens(configs: dict[str, dict[str, Any]],min_tokens: int) -> dict[str, dict[str, Any]]:
    """
    Filters and returns models whose max_tokens limit is >= min_tokens.
    Returns an empty dict if min_tokens exceeds all model capabilities.
    """
    return {
        model_name: config
        for model_name, config in configs.items()
        if config["max_tokens"] >= min_tokens
    }

def compare_temperatures(configs: dict[str, dict[str, Any]]) -> list[tuple[str, float]]:
    """
    Extracts the default temperature for each model and returns them
    sorted by temperature value from lowest temperature to most highest.
    """
    temp_list = [
        (model_name, config["temperature"]) 
        for model_name, config in configs.items()
    ]
    return sorted(temp_list, key=lambda item: item[1])

matched_openai = search_by_provider(MODELS_CONFIG, "openai")
for model, cfg in matched_openai.items():
    print(f"- {model} ({cfg['provider']})")

# Non-Existent Provider Test ('Meta')
matched_meta = search_by_provider(MODELS_CONFIG, "Meta")
print(f"Result: {matched_meta} (Length: {len(matched_meta)})")

#Filter by Max Tokens
capable_models = filter_by_max_tokens(MODELS_CONFIG, min_tokens=8000)
for model, cfg in capable_models.items():
    print(f"- {model}: max_tokens = {cfg['max_tokens']}")

print("\n=== 5. Compare Temperatures (Sorted Coolest to Hottest) ===")
sorted_temps = compare_temperatures(MODELS_CONFIG)
for rank, (model, temp) in enumerate(sorted_temps, start=1):
    print(f"{rank}. {model:<18} -> {temp}")
```
