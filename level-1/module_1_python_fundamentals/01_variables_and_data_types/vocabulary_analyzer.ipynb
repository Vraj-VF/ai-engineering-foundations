```python
import string

def read_text(path: str) -> str:
    """Opens a text file, reads its contents, and returns it as a single string."""
    with open(path, "r", encoding="utf-8") as file:
        return file.read()

def tokenize(text: str) -> list[str]:
    """Converts text to lowercase 1st then removes all punctuation, splits into words and we get a List of separated words"""
    text = text.lower()
    for p in string.punctuation:
        text = text.replace(p, ' ')
    return [word for word in text.split() if word]

def count_words(tokens: list[str]) -> dict[str, int]:
    """Counts the frequency of each word in the tokens list and we get a dictionary with word and it's frequency"""
    counts = {}
    for token in tokens:
        counts[token] = counts.get(token, 0) + 1
    return counts

def top_n(counts: dict[str, int], n: int = 10) -> list[tuple[str, int]]:
    """Returns the top n most frequent words as tuple to maintain the order and at last a List of sorted(high to low) numbers."""
    sorted_items = sorted(counts.items(), key=lambda item: item[1], reverse=True)
    return sorted_items[:n]

def avg_word_length(tokens: list[str]) -> float:
    """Calculates the average length of words."""
    if not tokens:  
        return 0.0
    return sum(len(t) for t in tokens) / len(tokens)

def richness(tokens: list[str]) -> float:
    """Calculates vocabulary richness (unique words / total words)."""
    if not tokens:  
        return 0.0
    return len(set(tokens)) / len(tokens)

sample_content = """Machine learning is a subset of artificial intelligence. 
Machine learning algorithms learn from data. 
Deep learning is a subset of machine learning.
AGI will rule the world."""

with open("sample.txt", "w", encoding="utf-8") as f:
    f.write(sample_content)

# 2. Read the text from the file using our new function
file_text = read_text("sample.txt")

# 3. Feed the file's text into our machines
words = tokenize(file_text)
word_counts = count_words(words)

top_10_words = top_n(word_counts, n=10)
average_length = avg_word_length(words)
vocab_richness = richness(words)

# 4. Print the results
print(f"Total Words: {len(words)}")
print(f"Average Word Length: {average_length:.2f} characters")
print(f"Vocabulary Richness: {vocab_richness:.2%} (ratio of unique words)")
print("\nTop 10 Words:")
for rank, (word, count) in enumerate(top_10_words, start=1):
    print(f"{rank}. '{word}' ({count} times)")
```
